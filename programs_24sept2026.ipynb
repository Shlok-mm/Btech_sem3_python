## wap to check if the number is Harshad number or not 



```python
num = int(input("Enter your number : "))
def harshad():
    temp = num
    sum = 0 
    while temp != 0 :
        rem = temp%10
        sum = sum + rem
        temp = temp //10
    if num % sum == 0  : 
        print("This is a harshad number !! ")
    else:
        print("This is not a harshad number !! ")


harshad()
```

## wap to make a tax calculator  



```python
grade = input("Enter your Grade [A to f]: ")
city = int(input("Enter your city class : "))

match grade :
    case 'A' : 
        basic_pay = 60000 
        allowance = 8000
    case 'B' : 
        basic_pay = 50000 
        allowance = 7000
    case 'C' : 
        basic_pay = 40000 
        allowance = 6000
    case 'D' : 
        basic_pay = 30000 
        allowance = 5000
    case 'E' : 
        basic_pay = 20000 
        allowance = 4000
    case 'F' : 
        basic_pay = 10000 
        allowance = 3000
    

match city :
    case 1 : 
        hra = basic_pay*0.30
    case 2 :
        hra = basic_pay*0.20
    case 3:
        hra = basic_pay*0.20

da = basic_pay * 0.50
pf = basic_pay * 0.11
transport = 900
ptax = 200 
gp = basic_pay + hra + da + allowance + transport - ptax - pf 
lpa = gp * 12 

print("Your grosspay is ",gp)
print("Your annual income is ",lpa)
```
