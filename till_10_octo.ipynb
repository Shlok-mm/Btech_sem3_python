# Tuple 
- Tuple is immutable.Tuple is read only version of list.
- If data os foxed and never changed then we can go for tuple.
- Insertion order is preserved.
- Duplicates are allowed.
- Hetrogeneous obkect are allowed.
- We can differentiate duplicate object by using index.
- Tuple supports both positive and negative indexing.
- Represent tuple element within parenthesis () with comma seprators.
- Parenthesis are optional but recommended.

```python
t = ()
type(t)
```

```python
t = ("string",)
type(t)
```

```python
t = (10,20,30)
print(t)
```

```python
t = tuple(range(0,50,2))
print(t)
```

#### Accessing the element of tuple 

1. By using index  

```python
t = (1,2,3,4,5,6,7,8)
print(t[1])
print(t[-5])
```

2. by using slicing operation 

```python
t = (1,2,3,4,5,6)
print(t[2:5])
print(t[2:100])
print(t[:100])
print(t[::2])
print(t[5:])
print(t[-5:-1])
print(t[-5:-1:-1])
print(t[5:1:1])
print(t[5:1:-1])
```
